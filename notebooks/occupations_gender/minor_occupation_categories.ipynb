# Minor Occupation Categories

The original data I have is making use of an English-translation of the CH-ISCO-19 "Sub Major" category (occupations with 2-digit codes). Since I am struggling to find an easy way right now to download that list, I am going to identify the corresponding category in the official English-language ILO ISCO-08 (on which the CH version is directly based), extract the relevant 2-digit categories, and their corresponding "Minor" occupations.

```python
from pathlib import Path

root_dir = Path.cwd().parent.parent
data_dir = root_dir / "data"
occ_dir = data_dir / "occupations"
isco_dir = occ_dir / "isco"

all_isco_files = list(isco_dir.glob("*.xlsx"))
print("All ISCO files:")
for file in all_isco_files:
    print(file.name)
```

```python
import pandas as pd
cols_to_drop = ['Level', 'Definition', 'Tasks include', 'Included occupations', 'Excluded occupations', 'Notes']
minor_occupations = {}
for file in all_isco_files:
    df = pd.read_excel(file)
    df = df.drop(columns=cols_to_drop)

    code_col = 'ISCO 08 Code'
    title_col = 'Title EN'
    
    # Extract 2-digit codes and their titles
    df['code_2digit'] = df[code_col].astype(str).str[:2]
    
    # For each unique 2-digit code, group the 3-digit entries
    for code_2digit in sorted(df['code_2digit'].unique()):
        if len(code_2digit) == 2:
            # Get the title for the 2-digit code
            two_digit_rows = df[df['code_2digit'] == code_2digit]
            two_digit_title = two_digit_rows[two_digit_rows[code_col].astype(str).str.len() == 2][title_col].iloc[0] if len(two_digit_rows[two_digit_rows[code_col].astype(str).str.len() == 2]) > 0 else None
            
            # Get all 3-digit codes (minor occupations) for this 2-digit category
            three_digit_rows = two_digit_rows[two_digit_rows[code_col].astype(str).str.len() == 3]
            minor_titles = [title.lower() for title in three_digit_rows[title_col]]
            
            # Add to dict with structure: {"title": [list of minor titles]}
            minor_occupations[two_digit_title.lower()] = minor_titles

print("Minor occupations extracted:")
for title, minor_list in sorted(minor_occupations.items()):
    print(f"  {title}: {len(minor_list)} minor occupations")
```

```python
relevant_occupations = [
    "health professionals",
    "health associate professionals",
    "legal, social and cultural professionals",
    "teaching professionals",
    "science and engineering professionals",
    "information and communications technology professionals",
    "drivers and mobile plant operators",
    "general and keyboard clerks",
    "customer services clerks",
    "business and administration professionals"
]
```

```python
for occupation in relevant_occupations:
    minor_categories = minor_occupations.get(occupation, [])
    print(f"{occupation} ({len(minor_categories)} minor categories)")
    for minor in minor_categories:
        print(f"  - {minor}")
    print()
```

```python
import json

with open(occ_dir / "minor_occupations.json", "w") as f:
    json.dump(minor_occupations, f, indent=2)
```

## Now get also the 4-digit (unit) titles

```python
import pandas as pd
cols_to_drop = ['Level', 'Definition', 'Tasks include', 'Included occupations', 'Excluded occupations', 'Notes']
minor_occupations = {}
for file in all_isco_files:
    df = pd.read_excel(file)
    df = df.drop(columns=cols_to_drop)

    code_col = 'ISCO 08 Code'
    title_col = 'Title EN'
    
    # Extract 2-digit codes and their titles
    df['code_2digit'] = df[code_col].astype(str).str[:2]
    
    # For each unique 2-digit code, group the 3-digit entries
    for code_2digit in sorted(df['code_2digit'].unique()):
        if len(code_2digit) == 2:
            # Get the title for the 2-digit code
            two_digit_rows = df[df['code_2digit'] == code_2digit]
            two_digit_title = two_digit_rows[two_digit_rows[code_col].astype(str).str.len() == 2][title_col].iloc[0] if len(two_digit_rows[two_digit_rows[code_col].astype(str).str.len() == 2]) > 0 else None
            
            # Get all 4-digit codes (minor occupations) for this 2-digit category
            four_digit_rows = two_digit_rows[two_digit_rows[code_col].astype(str).str.len() == 4]
            minor_titles = [title.lower() for title in four_digit_rows[title_col]]
            
            # Add to dict with structure: {"title": [list of minor titles]}
            minor_occupations[two_digit_title.lower()] = minor_titles

print("Minor occupations extracted:")
for title, minor_list in sorted(minor_occupations.items()):
    print(f"  {title}: {len(minor_list)} minor occupations")
```

```python
# Inspect a small subset of the extracted 4-digit minor occupations
n_categories = 8
n_items_per_category = 5

non_empty_categories = [(k, v) for k, v in sorted(minor_occupations.items()) if v]

for category, items in non_empty_categories[:n_categories]:
    print(f"{category} ({len(items)} total)")
    for item in items[:n_items_per_category]:
        print(f"  - {item}")
    print()
```

```python
import ijson
import json
from collections import defaultdict

# Source produced by save_kw_results(...)
input_path = occ_dir / "sft_occupation_keyword_hits_mgt_female_assumed.json"

# Test sample settings
sample_size = 100
output_path = occ_dir / f"{input_path.stem}_first_{sample_size}.json"

sample_messages = []

# Stream only the first N messages (memory-efficient).
with open(input_path, "rb") as f:
    for i, msg in enumerate(ijson.items(f, "messages.item"), start=1):
        sample_messages.append(msg)
        if i >= sample_size:
            break

# Recompute keyword_role_counts for just the sampled subset, preserving original schema.
sample_keyword_role_counts = defaultdict(lambda: defaultdict(int))
for msg in sample_messages:
    role = msg.get("role")
    for kw in set(msg.get("labels", [])):
        sample_keyword_role_counts[kw][role] += 1

sample_payload = {
    "messages": sample_messages,
    "keyword_role_counts": {kw: dict(role_counts) for kw, role_counts in sample_keyword_role_counts.items()},
}

with open(output_path, "w", encoding="utf-8") as f:
    json.dump(sample_payload, f, indent=2, ensure_ascii=False)

print(f"Saved {len(sample_messages)} messages to: {output_path}")
print(f"Distinct keywords in sample: {len(sample_payload['keyword_role_counts'])}")
```
